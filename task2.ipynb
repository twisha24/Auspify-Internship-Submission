```python
# step1
import pandas as pd
df = pd.read_csv("cleaned_dataset.csv")
```

```python
df.head(10)
```

```python
# step2
df["type"].value_counts()
```

```python

movies = (df["type"] == "Movie").sum()
tv_shows = (df["type"] == "TV Show").sum()
print("Total Movies:", movies)
print("Total TV Shows:", tv_shows)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# step3
sns.countplot(data=df,x="type")
plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Distribution")
content_percentage = df["type"].value_counts(normalize=True) * 100

print(content_percentage)

plt.show()
```

```python
# step4

plt.pie(
    content_percentage,
    labels=content_percentage.index,
    autopct="%1.1f%%",
    
)

plt.title("Percentage Distribution of Movies and TV Shows")
plt.show()
```

```python
# step5
print("Total Movies:", movies)
print("Total TV Shows:", tv_shows)
print("\nContent Percentage:")
print(content_percentage)
```
