```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("cleaned_dataset.csv")
```

```python
# step 1
df["release_year"].head(10)
```

```python
df["release_year"].dtype
```

```python
df["release_year"].isnull().sum()
```

```python
df_sorted = df.sort_values(by="release_year")
```

```python
df_sorted[["title", "release_year"]].head(10)
```

```python
# step2
yearly_releases = df["release_year"].value_counts().sort_index()

print(yearly_releases)
```

```python
# step3
yearly_growth = yearly_releases.diff()

print(yearly_growth)
```

```python
print("Highest releases year:", yearly_releases.idxmax())
print("Lowest releases year:", yearly_releases.idxmin())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12, 6))

plt.plot(yearly_releases.index, yearly_releases.values)

plt.title("Netflix Content Releases Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Releases")

plt.show()
```

```python
# step 5
top_5_years = yearly_releases.sort_values(ascending=False).head(5)

print(top_5_years)
```

```python
lowest_5_years = yearly_releases.sort_values().head(5)

print(lowest_5_years)
```

```python
### Business Insights – Year-wise Content Release Analysis

# - Netflix content releases increased significantly between 2016 and 2018.
# - The year 2018 recorded the highest number of releases, with 1,146 titles.
# - The years 2017 and 2019 both recorded 1,030 releases.
# - Content releases declined from 1,146 in 2018 to 953 in 2020.
# - Older release years have very few titles compared to recent years, indicating a greater representation of newer content in the dataset.
# Conclusion: The analysis shows that Netflix's content library contains a substantial number of titles released during 2016–2020, with 2018 being the peak year.
```
